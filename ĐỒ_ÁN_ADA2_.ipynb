```python
for name, df in datasets:
    print(f"KIỂM TRA BỘ DỮ LIỆU: {name.upper()}")
    print(f"{'='*30}")

    # Kiểm tra từng dòng có bị trùng lặp không?
    dup_rows_count = df.duplicated().sum()
    print(f"1. Số lượng dòng trùng lặp hoàn toàn: {dup_rows_count}")

    if dup_rows_count > 0:
        print(f" Các dòng trùng lặp (5 dòng đầu):")
        display(df[df.duplicated(keep=False)].head(5))
    else:
        print(" Không có dòng nào bị trùng lặp hoàn toàn.")

    # Kiểm tra từng cột có bị trùng lặp không?
    unique_cols_data = {}
    dup_data_cols = []

    for col in df.columns:
        col_data = tuple(df[col].iloc[:10000].values)

        if col_data in unique_cols_data:
            dup_data_cols.append((unique_cols_data[col_data], col))
        else:
            unique_cols_data[col_data] = col

    print(f"\n 2. Kiểm tra trùng lặp dữ liệu giữa các cột:")
    if dup_data_cols:
        print(f" Tìm thấy {len(dup_data_cols)} cặp cột giống hệt nhau:")
        display(pd.DataFrame(dup_data_cols, columns=['Cột Gốc', 'Cột Trùng']))
    else:
        print("Không tìm thấy cột nào có dữ liệu giống hệt nhau.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 4))

for i, col in enumerate(outlier_cols, 1):
    plt.subplot(1, 3, i)
    sns.boxplot(x=df_flights_active[col])
    plt.title(f'Boxplot của {col}')

plt.tight_layout()
plt.show()
```

```python
# Xử lý outliers bằng IQR (Capping)

# Tính ngưỡng IQR cho từng biến
outlier_bounds = {}

for col in outlier_cols:
    Q1 = df_flights_active[col].quantile(0.25)
    Q3 = df_flights_active[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_bounds[col] = {
        'lower': lower_bound,
        'upper': upper_bound
    }

outlier_bounds


# Capping outliers theo IQR
for col in outlier_cols:
    lower = outlier_bounds[col]['lower']
    upper = outlier_bounds[col]['upper']

    df_flights_active[col] = df_flights_active[col].clip(lower, upper)

```

```python
#So sánh trước và sau khi xử lý Outliers
plt.figure(figsize=(15, 4))

for i, col in enumerate(outlier_cols, 1):
    plt.subplot(1, 3, i)
    sns.boxplot(x=df_flights_active[col])
    plt.title(f'Boxplot {col} sau xử lý outliers')

plt.tight_layout()
plt.show()
```

**3.2 Xây dựng biến vận hành**

```python
# Biến 1: Tỷ lệ delay theo hãng hàng không
airline_delay_rate = df_flights_active.groupby('AIRLINE')['ARRIVAL_DELAY'].apply(
    lambda x: (x > 15).sum() / len(x)
).to_dict()
df_flights_active['AIRLINE_DELAY_RATE'] = df_flights_active['AIRLINE'].map(airline_delay_rate)

df_flights_active[['AIRLINE', 'AIRLINE_DELAY_RATE']].drop_duplicates().sort_values('AIRLINE_DELAY_RATE', ascending=False)
```

```python
# Biến 2: Tỷ lệ delay theo sân bay xuất phát
origin_delay_rate = df_flights_active.groupby('ORIGIN_AIRPORT')['ARRIVAL_DELAY'].apply(
    lambda x: (x > 15).sum() / len(x)
).to_dict()

df_flights_active['ORIGIN_DELAY_RATE'] = df_flights_active['ORIGIN_AIRPORT'].map(origin_delay_rate)
# Top 10 sân bay có tỷ lệ delay cao nhất
df_flights_active[['ORIGIN_AIRPORT', 'ORIGIN_DELAY_RATE']].drop_duplicates().sort_values(
    'ORIGIN_DELAY_RATE', ascending=False
).head(10)
```

```python
# Biến 3: Tỷ lệ delay theo sân bay đích
dest_delay_rate = df_flights_active.groupby('DESTINATION_AIRPORT')['ARRIVAL_DELAY'].apply(
    lambda x: (x > 15).sum() / len(x)
).to_dict()

df_flights_active['DEST_DELAY_RATE'] = df_flights_active['DESTINATION_AIRPORT'].map(dest_delay_rate)

# Top 10 sân bay đích có tỷ lệ delay cao nhất
df_flights_active[['DESTINATION_AIRPORT', 'DEST_DELAY_RATE']].drop_duplicates().sort_values(
    'DEST_DELAY_RATE', ascending=False
).head(10)
```

```python
# Biến 4: Khoảng cách bay (nếu chưa có)
# Đây là biến quan trọng để phân tích delay
if 'DISTANCE' in df_flights_active.columns:
    # Phân loại khoảng cách bay
    def categorize_distance(distance):
        if distance < 500:
            return 'SHORT'  # Ngắn
        elif distance < 1500:
            return 'MEDIUM'  # Trung bình
        else:
            return 'LONG'  # Dài

    df_flights_active['DISTANCE_CATEGORY'] = df_flights_active['DISTANCE'].apply(categorize_distance)

    # Xem phân bố
    print(df_flights_active['DISTANCE_CATEGORY'].value_counts())
    print('\nDelay trung bình theo loại khoảng cách:')
    print(df_flights_active.groupby('DISTANCE_CATEGORY')['ARRIVAL_DELAY'].mean())
```

```python
# Biến 5: Phân loại thời điểm trong ngày
def time_of_day(hour):
    if pd.isna(hour):
        return 'UNKNOWN'
    hour = int(hour / 100)  # Chuyển từ format HHMM sang HH
    if 5 <= hour < 12:
        return 'MORNING'  # Sáng (5AM-12PM)
    elif 12 <= hour < 17:
        return 'AFTERNOON'  # Chiều (12PM-5PM)
    elif 17 <= hour < 21:
        return 'EVENING'  # Tối (5PM-9PM)
    else:
        return 'NIGHT'  # Đêm (9PM-5AM)

df_flights_active['DEPARTURE_TIME_PERIOD'] = df_flights_active['DEPARTURE_TIME'].apply(time_of_day)

# Xem phân bố
print(df_flights_active['DEPARTURE_TIME_PERIOD'].value_counts())
print('\nDelay trung bình theo thời điểm:')
print(df_flights_active.groupby('DEPARTURE_TIME_PERIOD')['ARRIVAL_DELAY'].mean())
```

```python
# Biến 6: Tạo biến IS_DELAYED (target variable)
# Theo định nghĩa, chuyến bay delay nếu đến muộn > 15 phút
df_flights_active['IS_DELAYED'] = (df_flights_active['ARRIVAL_DELAY'] > 15).astype(int)

# Xem phân bố target
print('Phân bố biến mục tiêu IS_DELAYED:')
print(df_flights_active['IS_DELAYED'].value_counts())
print(f'\nTỷ lệ delay: {df_flights_active["IS_DELAYED"].mean()*100:.2f}%')
```

```python
# Biến 7: Tổng số delay reasons
delay_reason_cols = [
    'AIR_SYSTEM_DELAY',
    'SECURITY_DELAY',
    'AIRLINE_DELAY',
    'LATE_AIRCRAFT_DELAY',
    'WEATHER_DELAY'
]

# Đếm số loại nguyên nhân delay (>0) cho mỗi chuyến bay
df_flights_active['NUM_DELAY_REASONS'] = (df_flights_active[delay_reason_cols] > 0).sum(axis=1)

print('Phân bố số nguyên nhân delay:')
print(df_flights_active['NUM_DELAY_REASONS'].value_counts().sort_index())
```

```python
# Biến 8: Nguyên nhân delay chính
def get_main_delay_reason(row):
    reasons = {
        'AIR_SYSTEM': row['AIR_SYSTEM_DELAY'],
        'SECURITY': row['SECURITY_DELAY'],
        'AIRLINE': row['AIRLINE_DELAY'],
        'LATE_AIRCRAFT': row['LATE_AIRCRAFT_DELAY'],
        'WEATHER': row['WEATHER_DELAY']
    }

    # Lọc các nguyên nhân có giá trị > 0
    valid_reasons = {k: v for k, v in reasons.items() if v > 0}

    if not valid_reasons:
        return 'NONE'

    # Trả về nguyên nhân có thời gian delay lớn nhất
    return max(valid_reasons, key=valid_reasons.get)

df_flights_active['MAIN_DELAY_REASON'] = df_flights_active.apply(get_main_delay_reason, axis=1)

print('Phân bố nguyên nhân delay chính:')
print(df_flights_active['MAIN_DELAY_REASON'].value_counts())
```
